```python
from pathlib import Path
from netCDF4 import Dataset
import numpy as np

def extract_ssCCI(passes):
    BASE_DIR = Path("/nfs/DGFI141/A/SWH/CCI_v5/l2p-swh")
    MISSIONS_TO_PROCESS = ["sentinel-6_a"] 

    def iter_netcdf_files(mission_dir: Path):
        for year_dir in sorted([p for p in mission_dir.iterdir() if p.is_dir()]):
            for cycle_dir in sorted([p for p in year_dir.iterdir() if p.is_dir()]):
                for nc_path in sorted(cycle_dir.glob("*.nc")):
                    yield nc_path

    missions = [m for m in MISSIONS_TO_PROCESS if (BASE_DIR / m).is_dir()]
    file_list = []
    
    for mission in missions:
        mission_dir = BASE_DIR / mission
        print(f"=== Processing Mission: {mission} ===")
        file_count = 0

        for nc_path in iter_netcdf_files(mission_dir):
            try:
                # PERFORMANCE WARNING: Opening files over NFS is slow. 
                # Consider parsing the filename if it contains the pass number.
                with Dataset(nc_path, "r") as ds:
                    # Get attribute, defaulting to np.nan
                    pass_num = getattr(ds, "relative_pass_number", np.nan)
                    
                    # Cast to int to ensure it matches the types in your 'passes' list
                    if not np.isnan(pass_num) and int(pass_num) in passes:
                        file_count +=1
                        file_list.append(nc_path)

            except Exception as e:
                print(f"   [ERROR] It was not possible to read {nc_path.name}: {e}")
                continue

        print(f" -> Found {file_count} file .nc in total.")
        
    # FIX: The original function did not return anything
    return file_list
```

```python
import numpy as np
from netCDF4 import Dataset
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from pathlib import Path

BASE_DIR = Path("/nfs/DGFI141/A/SWH/CCI_v5/l2p-swh/sentinel-6_a")
                           
# Assuming 'files' is defined elsewhere in your notebook
swell_files = ["ESACCI-SEASTATE-L2P-SWH-Sentinel-6_A-20210306T141015-fv01.nc",
"ESACCI-SEASTATE-L2P-SWH-Sentinel-6_A-20210524T215828-fv01.nc",
"ESACCI-SEASTATE-L2P-SWH-Sentinel-6_A-20210910T234214-fv01.nc",
"ESACCI-SEASTATE-L2P-SWH-Sentinel-6_A-20210917T222827-fv01.nc",
"ESACCI-SEASTATE-L2P-SWH-Sentinel-6_A-20211003T185136-fv01.nc",
"ESACCI-SEASTATE-L2P-SWH-Sentinel-6_A-20211010T173748-fv01.nc",
"ESACCI-SEASTATE-L2P-SWH-Sentinel-6_A-20220101T003820-fv01.nc"]

wind_files = ["ESACCI-SEASTATE-L2P-SWH-Sentinel-6_A-20210908T002954-fv01.nc",
"ESACCI-SEASTATE-L2P-SWH-Sentinel-6_A-20210129T215002-fv01.nc",
"ESACCI-SEASTATE-L2P-SWH-Sentinel-6_A-20210228T154537-fv01.nc",
"ESACCI-SEASTATE-L2P-SWH-Sentinel-6_A-20210517T231215-fv01.nc",
"ESACCI-SEASTATE-L2P-SWH-Sentinel-6_A-20210521T224609-fv01.nc",
"ESACCI-SEASTATE-L2P-SWH-Sentinel-6_A-20210606T190918-fv01.nc",
"ESACCI-SEASTATE-L2P-SWH-Sentinel-6_A-20210812T054639-fv01.nc",
"ESACCI-SEASTATE-L2P-SWH-Sentinel-6_A-20211102T124709-fv01.nc"]

lat_min = -26
lat_max = -10
lon_min = 142
lon_max = 154

for filename in wind_files:
    found_paths = list(BASE_DIR.rglob(filename))
    if found_paths:
        file = found_paths[0] # This is now a complete Path object!
        try:
            with Dataset(file, "r") as ds:
                
                # 1. Pull latitude/longitude
                lat_raw = ds.variables["lat"][:]
                lon_raw = ds.variables["lon"][:]

                # 2. Extract the wave variables and flags
                swh_raw = ds.variables["swh_denoised"][:]
                shww_raw = ds.variables["era5_windwave_swh"][:]
                swh_quality = ds.variables["swh_quality_level"][:]
                
                # 3. Create the mask for valid, open ocean data
                # FIX 1: Typo 'lat-min' corrected to 'lat_min'
                mask = (swh_quality == 3) & (lat_raw > lat_min) & (lat_raw < lat_max) & (lon_raw > lon_min) & (lon_raw < lon_max)
                
                # 4. Filter data
                lat = lat_raw[mask]
                lon = lon_raw[mask]
                swh = swh_raw[mask]
                shww = shww_raw[mask]
                
                # FIX 2: Check if satellite actually passed over this region
                if len(lat) == 0:
                    print(f"Skipping {file.name} (or file): No data in region.")
                    continue
                    
                # FIX 3: Safely calculate global min and max using NumPy
                global_min = min(np.nanmin(shww), np.nanmin(swh))
                global_max = max(np.nanmax(shww), np.nanmax(swh))
                
                print(f"Plotting: {file}")    

                # 6. Set up the figure with 2 subplots (1 row, 2 columns)
                fig, ax = plt.subplots(1, 2, figsize=(20, 8), subplot_kw={'projection': ccrs.PlateCarree()})
                
                # --- LEFT PLOT: Wind Sea (shww) ---
                
                ax[0].set_extent([lon_min, lon_max, lat_min, lat_max], crs=ccrs.PlateCarree())
                
                # ADDED: This provides the ocean bathymetry background
                ax[0].stock_img() 
                
                ax[0].add_feature(cfeature.LAND, facecolor='lightgray', edgecolor='black', zorder=1)
                ax[0].add_feature(cfeature.COASTLINE, linewidth=1.5, zorder=2)
                
                gl0 = ax[0].gridlines(draw_labels=True, linestyle='--', alpha=0.5)
                gl0.top_labels = False
                gl0.right_labels = False

                # FIX 4: Apply the global_min and global_max to vmin and vmax
                sc0 = ax[0].scatter(lon, lat, c=shww, s=5, cmap='viridis', vmin=global_min, vmax=global_max, transform=ccrs.PlateCarree())
                ax[0].set_title("ERA5 Wind Sea (shww)")
                plt.colorbar(sc0, ax=ax[0], orientation='horizontal', pad=0.05, label='Wave Height (m)')

                # --- RIGHT PLOT: Calculated Swell ---
                
                ax[1].set_extent([lon_min, lon_max, lat_min, lat_max], crs=ccrs.PlateCarree())
                
                # ADDED: This provides the ocean bathymetry background here too
                ax[1].stock_img()
                
                ax[1].add_feature(cfeature.LAND, facecolor='lightgray', edgecolor='black', zorder=1)
                ax[1].add_feature(cfeature.COASTLINE, linewidth=1.5, zorder=2)
                
                gl1 = ax[1].gridlines(draw_labels=True, linestyle='--', alpha=0.5)
                gl1.top_labels = False
                gl1.right_labels = False
                
                # FIX 5: Apply the global_min and global_max to vmin and vmax here as well
                sc1 = ax[1].scatter(lon, lat, c=swh, s=5, cmap='plasma', vmin=global_min, vmax=global_max, transform=ccrs.PlateCarree())
                ax[1].set_title("SWH")
                plt.colorbar(sc1, ax=ax[1], orientation='horizontal', pad=0.05, label='Wave Height (m)')
                
                plt.tight_layout()
                plt.show()
                    
        except Exception as e:
            print(f"Error processing file {file}: {e}")
    else:
        print(f"WARNING: Could not find {filename} anywhere inside {BASE_DIR}")
```
